```python
# Your Phase 1 solution to the problem goes here.

# All of my imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

df = pd.read_csv("/content/sample_data/land_mines.csv")
```

```python
# 2. Split the sample 50/50 into training and test/validation sets.

eval_X_raw = df[["voltage", "height", "soil"]]
eval_y_raw = df["mine_type"]
(
   eval_X_train_raw,
   eval_X_valid_raw,
   eval_y_train,
   eval_y_valid
) = train_test_split(
    eval_X_raw,
    eval_y_raw,
    test_size=0.50,
    random_state=65
)

eval_scaler = MinMaxScaler()
eval_X_train_scaled = eval_scaler.fit_transform(eval_X_train_raw)
eval_X_valid_scaled = eval_scaler.transform(eval_X_valid_raw)
```

```python
# 3. Build a k -NN classifier
eval_ks = np.arange(1, 51)

eval_valid_sse = []
eval_train_sse = []

for eval_k in eval_ks:
    eval_candidate = KNeighborsClassifier(n_neighbors=int(eval_k))
    eval_candidate.fit(eval_X_train_scaled, eval_y_train)

    eval_valid_hat = eval_candidate.predict(eval_X_valid_scaled)

    eval_train_hat = eval_candidate.predict(eval_X_train_scaled)

    eval_valid_sse.append(np.sum(
        (eval_y_valid.to_numpy() - eval_valid_hat) ** 2
    ))
    eval_train_sse.append(np.sum(
        (eval_y_train.to_numpy() - eval_train_hat) ** 2
    ))
eval_k_star = int(eval_ks[np.argmin(eval_valid_sse)])
```

Q: Explain how you select k.

A: The way that I selected a k-value is by using the example from the slides where it shows how to find the k-value using a range.

```python
# 4. Print a confusion table for the optimal model, comparing predicted and actual class label on the test set.

eval_optimal_model = KNeighborsClassifier(n_neighbors=eval_k_star)
eval_optimal_model.fit(eval_X_train_scaled, eval_y_train)

eval_valid_prediction = eval_optimal_model.predict(eval_X_valid_scaled)

# Creating a confusion matrix
print(confusion_matrix(eval_y_valid, eval_valid_prediction))

# Creating the confusion table
confusion_table = pd.crosstab(
    eval_y_valid,
    eval_valid_prediction,
    rownames=['Actual'],
    colnames=['Predicted']
)

print(confusion_table)
```

Q: How accurate is it? Where is performance more or less accurate?

A: The values are accurately represented on both the confusion matrix and confusion table. It's more accurate when it's closer to 1 and less accurate when it's closer to 5.
