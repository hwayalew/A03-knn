1. What is the difference between regression and classification?

    The difference between regression and classification is it's purpose. Regression is used for a numeric target such as price and classification is used for a categorical target such as survival.

2. What is a confusion table? What does it help us understand about a model's performance?

    A confusion table/matrix is used to count predictions by actual class and predicted class using rows and columns (Predicted Negative, Predicted Positive, Actual Negative, Actual Positive). It helps us understand the relationship to determine the model's performance using the data.

3. What does the SSE quantify about a particular model?

    The SSE (Sum of Squared Error) quantifies the validation error based on a set using their observation and mean about a particular model.

4. What are overfitting and underfitting?

    Overfitting is calculated using the training data's average with smaller k-value. Underfitting is calculated using the training data's average with larger k-value.

5. Why does splitting the data into training and testing sets, and choosing $k$ by evaluating accuracy or SSE on the test set, improve model performance?

    Splitting the data into training and testing sets then choosing $k$ by evalutating accuracy or SSE on the test set improves model performance because it reducing the chances of overfitting or underfitting and the model from creating assumptions based on unseen data in hyperparameter (a setting chosen before fitting the model such as $k$).
